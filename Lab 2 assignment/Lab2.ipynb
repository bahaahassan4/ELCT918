```python
#trying to implement VGG16 CIFAR 10

import tensorflow as tf
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

print(f"Training images: {x_train.shape[0]}")
print(f"Test images: {x_test.shape[0]}")
print(f"Image shape: {x_train.shape[1:]}")

model = tf.keras.Sequential([
    ##first block contains 2 convs with 64 filters
    tf.keras.layers.Conv2D(64, (3,3),
        activation='relu',
        padding='same' ,## Keras automatically calculates and applies however much padding is needed so the output spatial size exactly matches the input spatial size (for stride 1)
        input_shape=(32,32,3)),
    ##32*32*64
    tf.keras.layers.Conv2D(64, (3,3),
        activation='relu',
        padding='same') ,
    ##32*32*64
    tf.keras.layers.MaxPooling2D((2,2)), #reducing feature map to quarter it's value
    ## 16*16*64


    ##second block contains 2 convs with 128 filters
    tf.keras.layers.Conv2D(128, (3,3),
        activation='relu',
        padding='same') ,
    ##16*16*128

    tf.keras.layers.Conv2D(128, (3,3),
        activation='relu',
        padding='same') ,
    ##16*16*128

    tf.keras.layers.MaxPooling2D((2,2)),
    ##8*8*128

    ##third block contains 3 convs with 256 filter
    tf.keras.layers.Conv2D(256, (3,3),
        activation='relu',
        padding='same') ,
    ##8*8*256
    tf.keras.layers.Conv2D(256, (3,3),
        activation='relu',
        padding='same') ,
    ##8*8*256
    tf.keras.layers.Conv2D(256, (3,3),
        activation='relu',
        padding='same') ,
    ##8*8*256
    tf.keras.layers.MaxPooling2D((2,2)),##4
    ##4*4*256

    ##fourth block contains 3 convs with 512 filters
    tf.keras.layers.Conv2D(512, (3,3),
        activation='relu',
        padding='same') ,
    ##4*4*512
    tf.keras.layers.Conv2D(512, (3,3),
        activation='relu',
        padding='same') ,
    ##4*4*512
    tf.keras.layers.Conv2D(512, (3,3),
        activation='relu',
        padding='same') ,
    ##4*4*512


    #tf.keras.layers.MaxPooling2D((2,2)),

    ##fifth block contains 3 convs with 512 filters
    tf.keras.layers.Conv2D(512, (3,3),
        activation='relu',
        padding='same') ,
    ##4*4*512
    tf.keras.layers.Conv2D(512, (3,3),
        activation='relu',
        padding='same') ,
    ##4*4*512
    tf.keras.layers.Conv2D(512, (3,3),
        activation='relu',
        padding='same') ,
    ##4*4*512

    #tf.keras.layers.MaxPooling2D((2,2)),

    tf.keras.layers.Flatten(),

    tf.keras.layers.Dense(4096, activation='relu'),##this layer is after flatten
                                                  ##so it's input is 4*4*512 = 8192
    tf.keras.layers.Dropout(0.5), ##0.5 means: during each training step, every neuron feeding into this point has a 50% chance of being
                                 #randomly set to zero (temporarily "dropped out") for that step only. A different random set gets dropped each batch.
                                 #at actual inference, dropout is automatically turned off
    tf.keras.layers.Dense(4096, activation='relu'),
    tf.keras.layers.Dropout(0.5),
    #tf.keras.layers.Dense(1024, activation='relu'),


    tf.keras.layers.Dense(10, activation='softmax')
##no dropout after the final Dense(10, softmax) layer — dropout is only applied to hidden layers,
##never to the output layer, since randomly zeroing out your actual class predictions would just corrupt the network's answer rather than regularize anything.
])


model.summary()

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'])

# added: validation_data so Keras evaluates on the test set at the end of every epoch,
# and capturing the return value of fit() into `history` so we have per-epoch numbers to plot
history = model.fit(x_train, y_train,
          epochs=10, batch_size=32,
          validation_data=(x_test, y_test))

model.evaluate(x_test, y_test)

# added: plot training vs validation accuracy and loss curves across epochs
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(history.history['accuracy'], label='Training accuracy')
ax1.plot(history.history['val_accuracy'], label='Validation accuracy')
ax1.set_title('Accuracy per epoch')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Accuracy')
ax1.legend()

ax2.plot(history.history['loss'], label='Training loss')
ax2.plot(history.history['val_loss'], label='Validation loss')
ax2.set_title('Loss per epoch')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Loss')
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
####
#trying to implement LeNet CIFAR 10
## but with the partial connections of C3 and S2
##


import tensorflow as tf
import matplotlib.pyplot as plt

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

print(f"Training images: {x_train.shape[0]}")
print(f"Test images: {x_test.shape[0]}")
print(f"Image shape: {x_train.shape[1:]}")

# Table I connectivity: which S2 feature maps (0-5) feed into each of C3's 16 feature maps
c3_connections = [
    (0,1,2), (1,2,3), (2,3,4), (3,4,5), (0,4,5), (0,1,5),            # 6 maps, 3 inputs each
    (0,1,2,3), (1,2,3,4), (2,3,4,5), (0,3,4,5), (0,1,4,5), (0,1,2,5), # 6 maps, 4 inputs each
    (0,1,3,4), (1,2,4,5), (0,2,3,5),                                  # 3 maps, 4 inputs each
    (0,1,2,3,4,5)                                                     # 1 map, all 6 inputs
]

##switched to Functional API (instead of Sequential) so C3 can connect to
##different subsets of S2's feature maps per the paper's Table I, rather
##than every C3 filter seeing all 6 input channels

inputs = tf.keras.Input(shape=(32,32,3))

##first block contains 1 conv with 6 filters
c1 = tf.keras.layers.Conv2D(6, (5,5),
    activation='relu')(inputs) #it should be tanh as the paper
##28*28*6

s2 = tf.keras.layers.AveragePooling2D((2,2))(c1) #reducing feature map to quarter it's value
## 14*14*6

##second block: 16 filters, each seeing only its assigned subset of S2's 6 maps (Table I)
c3_maps = []
for conn in c3_connections:
    selected = tf.keras.layers.Lambda(
        lambda x, idx=conn: tf.gather(x, idx, axis=-1)
    )(s2)
    branch = tf.keras.layers.Conv2D(1, (5,5), activation='relu')(selected)
    c3_maps.append(branch)
c3 = tf.keras.layers.Concatenate(axis=-1)(c3_maps)
##10*10*16

s4 = tf.keras.layers.AveragePooling2D((2,2))(c3)
##5*5*16

##third block contains 1 convs with 120 filters
c5 = tf.keras.layers.Conv2D(120, (5,5),
    activation='relu')(s4)
##1*1*120

flat = tf.keras.layers.Flatten()(c5)

f6 = tf.keras.layers.Dense(84, activation='relu')(flat)##this layer is after flatten
                                              ##so it's input is 1*1*120 = 120

outputs = tf.keras.layers.Dense(10, activation='softmax')(f6) ##replacement to the RBF (Euclidean Radial Basis Function)

model = tf.keras.Model(inputs=inputs, outputs=outputs)

model.summary()

model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=1e-4),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy'])

# added: validation_data so Keras evaluates on the test set at the end of every epoch,
# and capturing the return value of fit() into `history` so we have per-epoch numbers to plot
history = model.fit(x_train, y_train,
          epochs=10, batch_size=32,
          validation_data=(x_test, y_test))

model.evaluate(x_test, y_test)

# added: plot training vs validation accuracy and loss curves across epochs
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

ax1.plot(history.history['accuracy'], label='Training accuracy')
ax1.plot(history.history['val_accuracy'], label='Validation accuracy')
ax1.set_title('Accuracy per epoch')
ax1.set_xlabel('Epoch')
ax1.set_ylabel('Accuracy')
ax1.legend()

ax2.plot(history.history['loss'], label='Training loss')
ax2.plot(history.history['val_loss'], label='Validation loss')
ax2.set_title('Loss per epoch')
ax2.set_xlabel('Epoch')
ax2.set_ylabel('Loss')
ax2.legend()

plt.tight_layout()
plt.show()
```

```python
# trying to implement AlexNet CIFAR-10

import tensorflow as tf
import matplotlib.pyplot as plt


# ==============================
# Load CIFAR-10 Dataset
# ==============================

(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()


# Normalize pixel values
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0


print(f"Training images: {x_train.shape[0]}")
print(f"Test images: {x_test.shape[0]}")
print(f"Image shape: {x_train.shape[1:]}")


# ==============================
# AlexNet Architecture
# ==============================

model = tf.keras.Sequential([


    # ==============================
    # Conv Block 1
    # Original AlexNet:
    # 96 filters, 11x11, stride 4
    # ==============================

    tf.keras.layers.Conv2D(
        96,
        (11,11),
        strides=(4,4),
        activation='relu',
        padding='same',
        input_shape=(32,32,3)
    ),

    # output approximately:
    # 8*8*96


    tf.keras.layers.MaxPooling2D(
        (3,3),
        strides=(2,2),
        padding='same'
    ),

    # output:
    # 4*4*96



    # ==============================
    # Conv Block 2
    # Original AlexNet:
    # 256 filters, 5x5
    # ==============================

    tf.keras.layers.Conv2D(
        256,
        (5,5),
        activation='relu',
        padding='same'
    ),

    # 4*4*256


    tf.keras.layers.MaxPooling2D(
        (3,3),
        strides=(2,2),
        padding='same'
    ),

    # 2*2*256



    # ==============================
    # Conv Block 3
    # Original AlexNet:
    # 384 filters, 3x3
    # ==============================

    tf.keras.layers.Conv2D(
        384,
        (3,3),
        activation='relu',
        padding='same'
    ),

    # 2*2*384



    # ==============================
    # Conv Block 4
    # Original AlexNet:
    # 384 filters, 3x3
    # ==============================

    tf.keras.layers.Conv2D(
        384,
        (3,3),
        activation='relu',
        padding='same'
    ),

    # 2*2*384



    # ==============================
    # Conv Block 5
    # Original AlexNet:
    # 256 filters, 3x3
    # ==============================

    tf.keras.layers.Conv2D(
        256,
        (3,3),
        activation='relu',
        padding='same'
    ),

    # 2*2*256


    tf.keras.layers.MaxPooling2D(
        (3,3),
        strides=(2,2),
        padding='same'
    ),

    # 1*1*256



    # ==============================
    # Fully Connected Layers
    # ==============================

    tf.keras.layers.Flatten(),

    # input:
    # 1*1*256 = 256



    tf.keras.layers.Dense(
        4096,
        activation='relu'
    ),

    tf.keras.layers.Dropout(0.5),



    tf.keras.layers.Dense(
        4096,
        activation='relu'
    ),

    tf.keras.layers.Dropout(0.5),



    # CIFAR-10 has 10 classes

    tf.keras.layers.Dense(
        10,
        activation='softmax'
    )

])


model.summary()



# ==============================
# Compile Model
# ==============================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4
    ),

    loss='sparse_categorical_crossentropy',

    metrics=['accuracy']
)



# ==============================
# Training
# ==============================

history = model.fit(
    x_train,
    y_train,

    epochs=10,

    batch_size=32,

    validation_data=(x_test, y_test)
)



# ==============================
# Evaluation
# ==============================

model.evaluate(
    x_test,
    y_test
)



# ==============================
# Plot Accuracy and Loss Curves
# ==============================

fig, (ax1, ax2) = plt.subplots(
    1,
    2,
    figsize=(12,5)
)


# Accuracy

ax1.plot(
    history.history['accuracy'],
    label='Training accuracy'
)

ax1.plot(
    history.history['val_accuracy'],
    label='Validation accuracy'
)


ax1.set_title(
    'Accuracy per epoch'
)

ax1.set_xlabel(
    'Epoch'
)

ax1.set_ylabel(
    'Accuracy'
)

ax1.legend()


# Loss

ax2.plot(
    history.history['loss'],
    label='Training loss'
)

ax2.plot(
    history.history['val_loss'],
    label='Validation loss'
)


ax2.set_title(
    'Loss per epoch'
)

ax2.set_xlabel(
    'Epoch'
)

ax2.set_ylabel(
    'Loss'
)

ax2.legend()


plt.tight_layout()
plt.show()
```

```python
# trying to implement AlexNet CIFAR-100

import tensorflow as tf
import matplotlib.pyplot as plt

# ==============================
# Load CIFAR-100 Dataset
# ==============================
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar100.load_data()

# Normalize pixel values
x_train = x_train.astype('float32') / 255.0
x_test = x_test.astype('float32') / 255.0

print(f"Training images: {x_train.shape[0]}")
print(f"Test images: {x_test.shape[0]}")
print(f"Image shape: {x_train.shape[1:]}")

# ==============================
# AlexNet Architecture
# ==============================
model = tf.keras.Sequential([
    # ==============================
    # Conv Block 1
    # Original AlexNet:
    # 96 filters, 11x11, stride 4
    # ==============================
    tf.keras.layers.Conv2D(
        96,
        (11,11),
        strides=(4,4),
        activation='relu',
        padding='same',
        input_shape=(32,32,3)
    ),
    # output approximately:
    # 8*8*96

    tf.keras.layers.MaxPooling2D(
        (3,3),
        strides=(2,2),
        padding='same'
    ),
    # output:
    # 4*4*96

    # ==============================
    # Conv Block 2
    # Original AlexNet:
    # 256 filters, 5x5
    # ==============================
    tf.keras.layers.Conv2D(
        256,
        (5,5),
        activation='relu',
        padding='same'
    ),
    # 4*4*256

    tf.keras.layers.MaxPooling2D(
        (3,3),
        strides=(2,2),
        padding='same'
    ),
    # 2*2*256

    # ==============================
    # Conv Block 3
    # Original AlexNet:
    # 384 filters, 3x3
    # ==============================
    tf.keras.layers.Conv2D(
        384,
        (3,3),
        activation='relu',
        padding='same'
    ),
    # 2*2*384

    # ==============================
    # Conv Block 4
    # Original AlexNet:
    # 384 filters, 3x3
    # ==============================
    tf.keras.layers.Conv2D(
        384,
        (3,3),
        activation='relu',
        padding='same'
    ),
    # 2*2*384

    # ==============================
    # Conv Block 5
    # Original AlexNet:
    # 256 filters, 3x3
    # ==============================
    tf.keras.layers.Conv2D(
        256,
        (3,3),
        activation='relu',
        padding='same'
    ),
    # 2*2*256

    tf.keras.layers.MaxPooling2D(
        (3,3),
        strides=(2,2),
        padding='same'
    ),
    # 1*1*256

    # ==============================
    # Fully Connected Layers
    # ==============================
    tf.keras.layers.Flatten(),
    # input:
    # 1*1*256 = 256

    tf.keras.layers.Dense(
        4096,
        activation='relu'
    ),
    tf.keras.layers.Dropout(0.5),##0.5 means: during each training step, every neuron feeding into this point has a 50% chance of being
                                 #randomly set to zero (temporarily "dropped out") for that step only. A different random set gets dropped each batch.
                                 #at actual inference, dropout is automatically turned off

    tf.keras.layers.Dense(
        4096,
        activation='relu'
    ),
    tf.keras.layers.Dropout(0.5),

    # CIFAR-100 has 10 classes
    tf.keras.layers.Dense(
        100,
        activation='softmax'
    )
])


model.summary()



# ==============================
# Compile Model
# ==============================

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-5
    ),
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy', tf.keras.metrics.SparseTopKCategoricalAccuracy(k=5, name='top5_accuracy')]
)



# ==============================
# Training
# ==============================
# added: validation_data so Keras evaluates on the test set at the end of every epoch,
# and capturing the return value of fit() into `history` so we have per-epoch numbers to plot
history = model.fit(x_train,y_train,
          epochs=30, batch_size=32,
          validation_data=(x_test, y_test))

# ==============================
# Evaluation
# ==============================
model.evaluate(
    x_test,
    y_test
)

# ==============================
# Plot Accuracy and Loss Curves
# ==============================
fig, (ax1, ax2) = plt.subplots(
    1,
    2,
    figsize=(12,5)
)


# Accuracy
ax1.plot(
    history.history['accuracy'],
    label='Training accuracy'
)
ax1.plot(
    history.history['val_accuracy'],
    label='Validation accuracy'
)
ax1.set_title(
    'Accuracy per epoch'
)
ax1.set_xlabel(
    'Epoch'
)
ax1.set_ylabel(
    'Accuracy'
)
ax1.legend()


# Loss
ax2.plot(
    history.history['loss'],
    label='Training loss'
)
ax2.plot(
    history.history['val_loss'],
    label='Validation loss'
)
ax2.set_title(
    'Loss per epoch'
)
ax2.set_xlabel(
    'Epoch'
)
ax2.set_ylabel(
    'Loss'
)
ax2.legend()


plt.tight_layout()
plt.show()
```

# Overfitting Analysis

## Reason of Overfitting

- The model has a very large number of parameters compared to the CIFAR-100 dataset size.
- The two fully connected layers (`Dense(4096)`) contain millions of parameters, allowing the model to memorize training images instead of learning general features.
- The convolution architecture is also large for CIFAR-100, with many filters (96, 256, 384, 384, 256).
- The original AlexNet was designed for large images (224×224), while CIFAR-100 images are only 32×32, making the original architecture unnecessarily complex.
- The training accuracy keeps increasing while validation accuracy stops improving, and validation loss increases, indicating poor generalization.

# Recommended Solutions

- Reduce the model complexity:
  - Replace large fully connected layers (`Dense(4096)`) with smaller layers (e.g., `Dense(512)`).
  - Reduce the number of convolution filters.
  - Use smaller convolution kernels (3×3 instead of 11×11) for CIFAR-100.

- Apply regularization techniques:
  - Add Batch Normalization after convolution layers.
  - Increase Dropout rate if necessary.
  - Add L2 weight regularization.

- Improve training data:
  - Apply data augmentation (random flip, rotation, zoom, etc.) to increase dataset diversity.

- Use Early Stopping:
  - Stop training when validation loss stops improving and restore the best model weights.
